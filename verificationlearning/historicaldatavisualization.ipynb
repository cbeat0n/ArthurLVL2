```python
#Import dependencies
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set style, deprecated as of 09-07-2025 need to be tuned
sns.set(context="notebook", style="whitegrid")
```

```python
# --- Parameters ---
INPUT_CSV_PATH = 'verificationhistoricaldata/M1dataflow.csv'  # e.g. '/path/to/full_labeled_USD_JPY.csv'
OUTPUT_PNG_PATH = 'verificationhistoricaldocumentation/Visualizations/0125to0825M1.png'

# Column names
COL_TSEC = 't_sec'
COL_PRICE = 'value'
COL_VOLUME = 'volume'
COL_BUY = 'Buy'
COL_SELL = 'Sell'
COL_MARGIN = 'margin'
COL_THETA = 'theta'
COL_MU = 'mu'
COL_ACTIONHAT = 'ActionHat'
COL_TRUEACTION = 'TrueAction'

# If your TrueAction uses other labels, adjust order here
TRUEACTION_ORDER = [0, 1, 2]
```

```python
required_cols = [
    COL_TSEC, COL_PRICE, COL_VOLUME, COL_BUY, COL_SELL,
    COL_MARGIN, COL_THETA, COL_MU, COL_ACTIONHAT, COL_TRUEACTION
]

try:
    df  # see if df already exists
    _has_df = True
except NameError:
    _has_df = False

if not _has_df:
    if not INPUT_CSV_PATH:
        raise ValueError("No DataFrame named 'df' in memory and INPUT_CSV_PATH is empty. "
                         "Set INPUT_CSV_PATH or define 'df' before running.")
    df = pd.read_csv(INPUT_CSV_PATH)

missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"DataFrame is missing required columns: {missing}")

df = df.copy()
display(df.head(3))
print(f"Rows: {len(df):,}")
```

```python
def plot_bs_scatter(
    data: pd.DataFrame,
    x_col: str = COL_SELL,  # x = S
    y_col: str = COL_BUY,   # y = B
    hue_col: str = COL_TRUEACTION,
    hue_order = TRUEACTION_ORDER,
    output_path: str = OUTPUT_PNG_PATH,
    title: str = "B vs S — colored by TrueAction (0/1/2)",
    point_size: int = 14,
    alpha: float = 0.7
):
    """Create and save a scatterplot with x=S, y=B, hue=TrueAction (0/1/2)."""
    if hue_order is not None:
        data = data.copy()
        data[hue_col] = pd.Categorical(data[hue_col], categories=hue_order, ordered=True)

    fig, ax = plt.subplots(figsize=(7, 7))
    sns.scatterplot(
        data=data,
        x=x_col, y=y_col,
        hue=hue_col,
        hue_order=hue_order,
        s=point_size, alpha=alpha,
        linewidth=0, ax=ax
    )

    # Diagonal (B = S)
    min_lim = min(data[x_col].min(), data[y_col].min())
    max_lim = max(data[x_col].max(), data[y_col].max())
    ax.plot([min_lim, max_lim], [min_lim, max_lim], linestyle='--', linewidth=1, color='black')

    ax.set_xlabel('S (Sell)')
    ax.set_ylabel('B (Buy)')
    ax.set_title(title)
    ax.set_xlim(min_lim, max_lim)
    ax.set_ylim(min_lim, max_lim)
    ax.set_aspect('equal', adjustable='box')

    ax.legend(title=hue_col, bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.)
    plt.tight_layout()

    fig.savefig(output_path, dpi=160, bbox_inches='tight')
    plt.close(fig)
    print(f"Saved scatter to: {output_path}")
```

```python
#Jan-Aug M1
inputM1 = "verificationhistoricaldata/M1dataflow.csv"
outputM1 = 'verificationhistoricaldocumentation/Visualizations/0125to0825M1.png'
plot_bs_scatter(
    data = pd.read_csv(inputM1),
    output_path = outputM1,
    title = "B vs S — colored by TrueAction (0/1/2) || J-A25",
)
```
